### What did you know about your dataset?

The dataset consists of 5 linked tables covering vaccination coverage, disease incidence, reported cases, vaccine introduction status, and vaccine schedules across countries and WHO regions, spanning multiple years. Each file included a trailing metadata row (a timestamp of file creation) which was removed during cleaning. There are no duplicate rows in any table. The coverage table mixes country-level rows with World Bank income-group aggregate rows (GROUP = WB_LONG/WB_SHORT), which explains missing NAME values for those rows. Several numeric fields (TARGET_NUMBER, DOSES, COVERAGE, INCIDENCE_RATE, CASES) have substantial missingness, reflecting inconsistent reporting across countries/years rather than data errors — these will be retained as NULLs rather than imputed.

### Variables Description

Coverage table — GROUP: whether the row is a country or an aggregate (income group); CODE: ISO3 country code or aggregate code; NAME: country name (blank for aggregates); YEAR: reporting year; ANTIGEN/ANTIGEN_DESCRIPTION: vaccine code and full name; COVERAGE_CATEGORY/_DESCRIPTION: whether coverage is administrative or official estimate; TARGET_NUMBER: population targeted for that dose; DOSES: doses administered; COVERAGE: % of target population vaccinated.

Incidence table — same GROUP/CODE/NAME/YEAR structure; DISEASE/DISEASE_DESCRIPTION: disease code and name; DENOMINATOR: the population basis the rate is calculated against (e.g. per 1,000,000 population); INCIDENCE_RATE: cases per that denominator.

Cases table — same geography/year columns; DISEASE/DISEASE_DESCRIPTION: as above; CASES: raw reported case count (not a rate).

Vaccine Introduction table — ISO_3_CODE/COUNTRYNAME/WHO_REGION/YEAR: geography and time; DESCRIPTION: vaccine name; INTRO: Yes/No whether that vaccine was introduced into the country's national program that year.

Vaccine Schedule table — same geography/year columns; VACCINECODE/VACCINE_DESCRIPTION: vaccine identity; SCHEDULEROUNDS: which dose number in the schedule; TARGETPOP/_DESCRIPTION: who the dose targets (e.g. infants, women); GEOAREA: national vs subnational rollout; AGEADMINISTERED: age/timing of the dose; SOURCECOMMENT: free-text notes on the source.
